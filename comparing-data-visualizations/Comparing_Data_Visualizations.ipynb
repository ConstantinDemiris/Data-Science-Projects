# Comparing Data Visualizations

This notebook uses `countries.csv` to compare life expectancy in Asia and Africa. It includes a time-series comparison across all available years with 1952 highlighted, followed by separate 1952 histograms created with pyplot's `hist()` and `subplot()` functions.

## 1. Import the required libraries

Pandas is used to load, filter, and summarize the country data. NumPy supplies consistent histogram boundaries, and Matplotlib's pyplot module creates the time series and histogram subplots.

```python
# Import pandas for working with the countries CSV file.
import pandas as pd

# Import NumPy for shared histogram bin boundaries.
import numpy as np

# Import pyplot for time-series plots, histograms, and subplots.
from matplotlib import pyplot as plt
```

## 2. Load `countries.csv`

The CSV file is loaded into a pandas DataFrame named `data`. The first rows, shape, columns, and missing-value counts are displayed to verify the source before analysis.

```python
# Read the course dataset stored in the same folder as this notebook.
data = pd.read_csv("countries.csv")

# Inspect the dataset to confirm that it loaded correctly.
print(data.head())
print("\nDataset shape:", data.shape)
print("Columns:", data.columns.tolist())
print("\nMissing values by column:")
print(data.isna().sum())
```

## 3. Filter Asia and Africa for 1952

The dataset is first filtered to 1952. Separate DataFrames are then created for Asia and Africa so their life-expectancy distributions can be compared using the same measurement scale.

```python
# Keep only records from the required year.
data_1952 = data[data["year"] == 1952].copy()

# Create separate 1952 datasets for the two continents being compared.
asia_1952 = data_1952[data_1952["continent"] == "Asia"].copy()
africa_1952 = data_1952[data_1952["continent"] == "Africa"].copy()

# Confirm how many countries are represented in each comparison group.
print("Asian countries in 1952:", asia_1952["country"].nunique())
print("African countries in 1952:", africa_1952["country"].nunique())
```

## 4. Calculate 1952 comparison statistics

Mean and median life expectancy summarize the center of each continent's distribution. Standard deviation describes how widely individual country values vary around the mean.

```python
# Build a concise statistical comparison for Asia and Africa in 1952.
comparison_1952 = pd.DataFrame({
    "Asia": {
        "Country Count": asia_1952["country"].nunique(),
        "Mean Life Expectancy": asia_1952["lifeExpectancy"].mean(),
        "Median Life Expectancy": asia_1952["lifeExpectancy"].median(),
        "Standard Deviation": asia_1952["lifeExpectancy"].std(),
    },
    "Africa": {
        "Country Count": africa_1952["country"].nunique(),
        "Mean Life Expectancy": africa_1952["lifeExpectancy"].mean(),
        "Median Life Expectancy": africa_1952["lifeExpectancy"].median(),
        "Standard Deviation": africa_1952["lifeExpectancy"].std(),
    },
})

print(comparison_1952.round(2))
```

## 5. Create a time series comparing Asia and Africa

A time series requires multiple years. The code calculates the average life expectancy for each continent in every available year and draws both trends on the same chart. The 1952 observations are highlighted to connect the time series to the required histogram comparison.

```python
# Keep Asia and Africa, then calculate average life expectancy by year and continent.
life_expectancy_series = (
    data[data["continent"].isin(["Asia", "Africa"])]
    .groupby(["year", "continent"], as_index=False)["lifeExpectancy"]
    .mean()
)

# Draw one line for each continent so their change over time can be compared.
plt.figure(figsize=(10, 6))
for continent, color in [("Asia", "royalblue"), ("Africa", "darkorange")]:
    continent_series = life_expectancy_series[
        life_expectancy_series["continent"] == continent
    ]
    plt.plot(
        continent_series["year"],
        continent_series["lifeExpectancy"],
        marker="o",
        linewidth=2,
        color=color,
        label=continent,
    )

# Highlight the required year with a vertical reference line.
plt.axvline(1952, color="gray", linestyle="--", label="1952 comparison")
plt.title("Average Life Expectancy in Asia and Africa Over Time")
plt.xlabel("Year")
plt.ylabel("Average Life Expectancy (Years)")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## 6. Compare the 1952 distributions with `hist()` and `subplot()`

The two histograms use identical bins and x-axis limits, making their shapes directly comparable. `subplot(2, 1, 1)` places Asia on the top row, while `subplot(2, 1, 2)` places Africa below it.

```python
# Use identical five-year intervals for both histograms.
histogram_bins = np.arange(20, 71, 5)

# Create the figure that will contain two vertically stacked subplots.
plt.figure(figsize=(10, 8))

# Draw the Asia histogram in the first subplot.
plt.subplot(2, 1, 1)
plt.hist(
    asia_1952["lifeExpectancy"],
    bins=histogram_bins,
    edgecolor="black",
    color="royalblue",
    alpha=0.80,
)
plt.axvline(
    asia_1952["lifeExpectancy"].mean(),
    color="red",
    linestyle="--",
    label="Mean",
)
plt.title("Distribution of Life Expectancy in 1952")
plt.ylabel("Asia: Number of Countries")
plt.xlim(20, 70)
plt.legend()

# Draw the Africa histogram in the second subplot using the same scale.
plt.subplot(2, 1, 2)
plt.hist(
    africa_1952["lifeExpectancy"],
    bins=histogram_bins,
    edgecolor="black",
    color="darkorange",
    alpha=0.80,
)
plt.axvline(
    africa_1952["lifeExpectancy"].mean(),
    color="red",
    linestyle="--",
    label="Mean",
)
plt.xlabel("Life Expectancy (Years)")
plt.ylabel("Africa: Number of Countries")
plt.xlim(20, 70)
plt.legend()

plt.tight_layout()
plt.show()
```

## 7. Interpretation

In 1952, the average life expectancy was approximately **46.31 years in Asia** and **39.14 years in Africa**. The Asian distribution is shifted toward higher life-expectancy values, although both continents show substantial variation among countries. The time-series chart shows that average life expectancy increased in both regions after 1952, while Asia maintained the higher continental average throughout the available period.
